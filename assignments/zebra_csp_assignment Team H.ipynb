# CSP Programming Assignment Simplified

## Zebra Puzzle as a Constraint Satisfaction Problem (CSP)

## 1. Problem Description
We are given a classic logic puzzle involving five houses. Each house has a unique:
- Color
- Nationality of the owner
- Drink
- Candy
- Pet

The goal is to determine:
- Who owns the zebra?
- Who drinks water?


## 2. CSP Formulation

### Variables
Each attribute value is treated as a variable:
- Nationalities: Englishman, Spaniard, Norwegian, Ukrainian, Japanese
- Colors: Red, Green, Ivory, Yellow, Blue
- Drinks: Coffee, Tea, Milk, Orange Juice, Water
- Candies: Hershey, KitKat, Smarties, Snickers, MilkyWay
- Pets: Dog, Fox, Snails, Horse, Zebra

### Domains
Each variable takes a value in {1,2,3,4,5}, representing house positions.

### Constraints
- Englishman = Red
- Spaniard = Dog
- Norwegian = 1
- Green = Ivory + 1
- Hershey next to Fox
- KitKat = Yellow
- Norwegian next to Blue
- Smarties = Snails
- Snickers = Orange Juice
- Ukrainian = Tea
- Japanese = MilkyWay
- KitKat next to Horse
- Coffee = Green
- Milk = 3


## 3. Representation Choice

We use the **attribute-based representation**, where each variable represents a specific attribute value and its domain is the house number.

### Why this representation?
- Constraints become simple equations (e.g., Englishman = Red)
- Efficient constraint propagation
- Standard approach in CSP solving


```python
import itertools

houses = [1, 2, 3, 4, 5]

def next_to(a, b):
    return abs(a - b) == 1

def right_of(a, b):
    return a == b + 1
```

## 4. CSP Solver

```python
solutions = []

for nat in itertools.permutations(houses):
    Englishman, Spaniard, Norwegian, Ukrainian, Japanese = nat

    if Norwegian != 1:
        continue

    for col in itertools.permutations(houses):
        Red, Green, Ivory, Yellow, Blue = col

        if Englishman != Red:
            continue
        if not right_of(Green, Ivory):
            continue
        if not next_to(Norwegian, Blue):
            continue

        for dri in itertools.permutations(houses):
            Coffee, Tea, Milk, OrangeJuice, Water = dri

            if Coffee != Green:
                continue
            if Ukrainian != Tea:
                continue
            if Milk != 3:
                continue

            for can in itertools.permutations(houses):
                Hershey, KitKat, Smarties, Snickers, MilkyWay = can

                if KitKat != Yellow:
                    continue
                if Snickers != OrangeJuice:
                    continue
                if Japanese != MilkyWay:
                    continue

                for pet in itertools.permutations(houses):
                    Dog, Fox, Snails, Horse, Zebra = pet

                    if Spaniard != Dog:
                        continue
                    if Smarties != Snails:
                        continue
                    if not next_to(Hershey, Fox):
                        continue
                    if not next_to(KitKat, Horse):
                        continue

                    solutions.append((nat, col, dri, can, pet))
```

## 5. Results

```python
solution = solutions[0]

nat, col, dri, can, pet = solution

names_nat = ["Englishman", "Spaniard", "Norwegian", "Ukrainian", "Japanese"]
names_pet = ["Dog", "Fox", "Snails", "Horse", "Zebra"]
names_drink = ["Coffee", "Tea", "Milk", "Orange Juice", "Water"]

zebra_house = pet[names_pet.index("Zebra")]
water_house = dri[names_drink.index("Water")]

owner_zebra = names_nat[nat.index(zebra_house)]
drinks_water = names_nat[nat.index(water_house)]

print("The Zebra is owned by the", owner_zebra)
print("The Water is drunk by the", drinks_water)
```

## Abstract:
The Zebra Puzzle is a constraint satisfaction problem, consisting of five houses of differing attributes such as nationality of the inhabitants, color, preferences, and animals. The goal is to correctly determine which house contains the zebra and which inhabitant likes to drink water by applying the given logical constraints. By using CSP techniques, we were able to model the puzzle using variables, domains and constraints, enabling us to compare multiple representations by focusing on the houses, the attributes, and the binary constraints and evaluating the different compuational implications of each representation including efficiency, scalability, readibility and ease of implementation. In the end, we have concluded that while multiple CSP formulations exist, some have different applicability for different problems due to their varying search complexity, constraint propagation, and reasoning.

## Introduction:
The puzzle describes five houses in a row, each with a distinct color, nationality of the inhabitant, preferred drink, preferred sweet, and pet animal. The objective was to determine which resident owns the zebra and which one likes to drink water, based on the information given. The assignment's purpose was not only to solve the puzzle, but also demonstrate how a logical problem can be translated into a computational model. This formulation is important because the ways constraints and variables are presented vastly affect readability and efficiency.

## Theory:
The puzzle consists of three parts: a set of variables, a domain for each variable, and a set of constraints. The solution is an assignment of values for all all variables in a way that every given constraint is satisfied. Here, the variables correspond to different attribute values for each house, with the domain of each variable being the set of house positions, answering the question of "to which house does this apply to?"

The clue set thus becomes a collection of equality and adjacency constraints. Because this is a finite-domain constraint problem, it can be solved by systematic search combined with constraint checks. In our case, we used permutation-based search, equivalent to trying consistent assignements category-by-category while rejecting contradictory assignments. 

## Discussion:
Using the attribute-based representation, each valye from each category is treated as a variable whose value is the house in which it appears in order, which is helpful because the Zebra puzzle is fundamentally about relationships between attributes as opposed to objects with internal structures, and thus the constraints are easy to express and verify.

Using iterools.permutations to list all possible house assignments for each category, then applying all the constraints as early as possible to discard invalid combinations, allows us to respect the uniqueness of each category using permutations, reduce the search space, and return the unique complete solution needed for the exercise.

In the end, the final answer resulted in the conclusion that the zebra is owned by the Japanese, who lives in the fifth house, and water is drunk by the Norwegian, who lives in the first house.

## Conclusion:

The Zebra puzzle is a classic example of a CSP because it combines finite domains with relational constraints. The implementation we concluded with solves the puzzle using an attribute-based representation in which each attribute is assigned a house number, making the constraints direct, compact, and easy to evaluate. Though other representations are possible, the attribute-based approach was the one we concluded was the most suitable because it supports clear reasoning and efficiency.
